Файл weather_data_russia_2014_2024.xlsx будет содержать таблицу с колонками:

 - date: Дата измерения.
 - tavg: Средняя температура за день.
 - tmin: Минимальная температура за день.
 - tmax: Максимальная температура за день.
 - prcp: Количество осадков.
 - snow: Высота снежного покрова.
 - wdir: Направление ветра.
 - wspd: Скорость ветра.
 - wpgt: Порывы ветра.
 - pres: Атмосферное давление.
 - tsun: Продолжительность солнечного сияния.

```python
import requests
import pandas as pd
from datetime import datetime

# Ваш API-ключ
API_KEY = 'ваш_api_ключ'

# Функция для получения данных о погоде за указанный период
def get_weather_data(lat, lon, start_date, end_date):
    url = f"https://api.meteostat.net/v2/point/daily"
    params = {
        'lat': lat,
        'lon': lon,
        'start': start_date,
        'end': end_date,
        'tz': 'Europe/Moscow',
        'key': API_KEY
    }
    
    response = requests.get(url, params=params)
    if response.status_code == 200:
        return response.json()['data']
    else:
        print(f"Ошибка: {response.status_code}, {response.text}")
        return None

# Основная функция для сбора данных за период
def collect_weather_data(lat, lon, start_year, end_year):
    all_data = []
    
    for year in range(start_year, end_year + 1):
        start_date = f"{year}-01-01"
        end_date = f"{year}-12-31"
        print(f"Сбор данных за {year}...")
        data = get_weather_data(lat, lon, start_date, end_date)
        if data:
            all_data.extend(data)
    
    return all_data

# Сохранение данных в Excel
def save_to_excel(data, filename):
    df = pd.DataFrame(data)
    df.to_excel(filename, index=False)
    print(f"Данные сохранены в файл {filename}")

# Параметры сбора данных
LAT = 55.7558  # Широта (Москва)
LON = 37.6173  # Долгота (Москва)
START_YEAR = 2014
END_YEAR = 2024
OUTPUT_FILE = "weather_data_russia_2014_2024.xlsx"

# Сбор данных
weather_data = collect_weather_data(LAT, LON, START_YEAR, END_YEAR)

# Сохранение данных в Excel
save_to_excel(weather_data, OUTPUT_FILE)
```
